```python
dtype = {
    "Libellé civilité d'exercice": "str",
    "Libellé savoir-faire": "str",
    "Code postal (coord. structure)": "str",
}

df = pd.read_csv(
    "./data/praticiens.txt",
    sep="|",
    header=0,
    dtype=dtype,
    usecols=[
        "Libellé civilité d'exercice",
        "Libellé civilité",
        "Nom d'exercice",
        "Prénom d'exercice",
        "Libellé profession",
        "Libellé savoir-faire",
        "Code mode exercice",
        "Numéro Voie (coord. structure)",
        "Libellé type de voie (coord. structure)",
        "Libellé Voie (coord. structure)",
        "Code postal (coord. structure)",
        "Libellé commune (coord. structure)",
        "Libellé secteur d'activité",
    ],
)
```

```python
medecins = df[df["Libellé profession"] == "Médecin"]
```

```python
medecins["Libellé savoir-faire"].unique()
```

```python
generalistes = medecins[
    medecins["Libellé savoir-faire"].isin(
        [
            "Médecine Générale",
            "Qualifié en Médecine Générale",
            "Spécialiste en Médecine Générale",
        ]
    )
]
```

```python
generalistes.info()
```

```python
generalistes = generalistes.dropna(
    subset=["Code postal (coord. structure)", "Nom d'exercice"]
)
```

```python
generalistes.info()
```

```python
generalistes[generalistes["Code mode exercice"] == "L"][
    "Libellé secteur d'activité"
].unique()
```

```python
secteurs_ville = [
    "Cabinet individuel",
    "Cabinet de groupe",
    "Exercice en Société",
    "Centres de santé",
]
generalistes_ville = generalistes[
    (generalistes["Code mode exercice"] == "L")
    & generalistes["Libellé secteur d'activité"].isin(secteurs_ville)
]
```

```python
generalistes_ville.info()
```

```python
generalistes_ville_idf = generalistes_ville[
    generalistes_ville["Code postal (coord. structure)"].str.startswith(
        ("75", "92", "93", "94", "77", "91", "78", "95")
    )
]
```

```python
generalistes_ville_creuse = generalistes_ville[
    generalistes_ville["Code postal (coord. structure)"].str.startswith(
        "23"
    )
]
```

```python
mes_generalistes = generalistes_ville_creuse
```

```python
mes_generalistes.count()
```

```python
mes_generalistes["Numéro Voie (coord. structure)"] = (
    mes_generalistes["Numéro Voie (coord. structure)"].astype("Int64").astype(str)
)
```

```python
mes_generalistes.dtypes
```

```python
mes_generalistes["Numéro Voie (coord. structure)"] = mes_generalistes[
    "Numéro Voie (coord. structure)"
].fillna(value="")
```

```python
mes_generalistes["Libellé Voie (coord. structure)"] = mes_generalistes[
    "Libellé Voie (coord. structure)"
].fillna("")
```

```python
mes_generalistes["adresse"] = (
    mes_generalistes["Numéro Voie (coord. structure)"]
    + " "
    + mes_generalistes["Libellé Voie (coord. structure)"]
    + " "
    + mes_generalistes["Code postal (coord. structure)"]
    + " "
    + mes_generalistes["Libellé commune (coord. structure)"]
)
```
